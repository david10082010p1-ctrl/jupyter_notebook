
# Uso da biblioteca json para manipular arquivos JSON em Python

## 1. modificar arquivos json

```python
import json 
from datetime import datetime

with open('dict.json', 'r', encoding = "utf-8") as file:
    data = json.load(file)

data['modification_at'] = datetime.now().day

with open('dict.json', 'w',encoding = "utf-8") as file:
    json.dump(data,file, indent=4)
print(data)
```

### 2. deletar objetos em arquivos Json

esta estrutura é fundamental para fazermos deleções em objetos indesejados, para substuiirmos posteriormente


```python
import json 

with open('dict.json', 'r') as file:
    data = json.load(file)

if 'modification_at' in data:
    del data['modification_at']


print(data)
```
